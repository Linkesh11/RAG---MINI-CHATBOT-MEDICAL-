```python
from langchain_community.document_loaders import PyPDFLoader

from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_community.vectorstores import FAISS
from langchain_huggingface import HuggingFaceEmbeddings

```

```python
loader = PyPDFLoader(r"D:/numpy/final_project/Common Eye Diseases and their Management.pdf")

docs = loader.load()
```

```python
print(len(docs))
print(docs[0].page_content[:500])
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

chunks = text_splitter.split_documents(docs)
```

```python
print("Total Chunks:", len(chunks))
```

```python
print(chunks[0].page_content)
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vector = embeddings.embed_query(chunks[0].page_content)

print(len(vector))
```

```python
db = FAISS.from_documents(
    documents=chunks,
    embedding=embeddings
)

print(db)
```

```python
docs_dict = db.docstore._dict

print("Total Stored Documents:", len(docs_dict))
```

```python
retriever = db.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 3}
)

```

```python
question = "What are the Physiology of the Eye?"

results = retriever.invoke(question)

results
```

```python
from transformers import pipeline
import torch

pipe = pipeline(
    "text-generation",
    model="Qwen/Qwen2.5-1.5B-Instruct",
    device=0 if torch.cuda.is_available() else -1,
    max_new_tokens=300,
    temperature=0,
    do_sample=False,
    return_full_text=False
)


from langchain_huggingface import HuggingFacePipeline

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
def eye_rag(question):

    docs = retriever.invoke(question)

    context = "\n\n".join(doc.page_content for doc in docs)

    prompt = f"""
You are a medical assistant.

STRICT RULES:
1. Answer ONLY using the provided context.
2. Do NOT use any external knowledge.
3. Do NOT infer or assume missing information.
4. If the answer is not explicitly present in the context, reply exactly:
"I don't have enough information."
5. Do not add introductions or conclusions.

Context:
{context}

Question:
{question}

Answer:
"""

    answer = llm.invoke(prompt)

    return answer
```

```python
question = input("Enter the Question")

answer = eye_rag(question)

print(answer)
```
